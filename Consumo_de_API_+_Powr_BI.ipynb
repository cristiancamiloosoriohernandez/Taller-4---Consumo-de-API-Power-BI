# CONSULTA DE API

### 1. Importar la librería y definir la URL de la API

Usaremos la librería `requests`. Si deseas consultar una API que requiere autenticación o parámetros, puedes incluirlos en la consulta. En este ejemplo, usaremos una API pública gratuita que devuelve datos de usuarios de prueba.

```python
import requests

# URL de la API de datos.gov.co proporcionada
url = 'https://www.datos.gov.co/api/v3/views/68pj-ac7b/query.json?query=SELECT%20departamento%2C%20ciudad%2C%20placa_vehiculo%2C%20fecha%2C%20codigo_infraccion%20ORDER%20BY%20fecha%20DESC'

# Realizar la petición GET
respuesta = requests.get(url)
```

### 2. Verificar el estado y mostrar los resultados

Es importante verificar si la petición fue exitosa (código de estado `200`) antes de procesar los datos.

```python
if respuesta.status_code == 200:
    # Convertir la respuesta a formato JSON
    datos = respuesta.json()
    print("¡Consulta exitosa a datos.gov.co!")
    # Mostramos los primeros 5 elementos para no saturar la pantalla si son muchos datos
    display(datos[:5] if isinstance(datos, list) else datos)
else:
    print(f"Error al consultar la API. Código de estado: {respuesta.status_code}")
    print(respuesta.text)
```

```python
import pandas as pd

df = pd.DataFrame(datos)
df.head()
```
